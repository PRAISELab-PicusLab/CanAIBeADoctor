```python
from openai import OpenAI
import pandas as pd
import time
import os

def clean_console():
    try:
        # Prova a usare clear_output se siamo in ambiente IPython (Colab, Jupyter)
        from IPython.display import clear_output
        clear_output(wait=True)
    except ImportError:
        # Se non siamo in IPython, prova clear normale
        import os
        os.system('cls' if os.name == 'nt' else 'clear')


def update_dataframe(index, answer):
    """
    Aggiorna il DataFrame `df` aggiungendo la risposta nella colonna 'Answer_medico_caring' corrispondente all'indice.
    """
    global df
    df.loc[index, 'Answer_medico_caring_LLM_PALMYRA'] = answer

# Configura il client OpenAI58299729
client = OpenAI(
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=""
)

# Parametri iniziali
start_index = 9729
num_rows = 10000
save_every = 100  # salva ogni 100 righe
output_path = '/content/drive/MyDrive/lavoro/Ricerca/CHI/df1csv_caring.csv'

count = 0

for index in range(start_index, start_index + num_rows):
    try:
        # Preleva la risposta originale dal DataFrame
        original_answer = df.loc[index, 'Answer_medico']
        question =df.loc[index, 'question']

        # Salta se la risposta è vuota o NaN
        if pd.isna(original_answer) or original_answer.strip() == "":
            continue

        # Costruisci il prompt
        prompt_template = f'''[INST] <<SYS>>
You are a compassionate and trustworthy medical doctor.

Your task is to rephrase the provided medical answer to a certain question to make it easier to understand for someone without a medical background.

Keep the meaning exactly the same, but use a warm, supportive, and caring tone. Speak simply and clearly. Imagine you are explaining it kindly to a patient.

Avoid lists, bullet points, or numbered items. Use short sentences and everyday words. Do not add new information.

Follow the Flesch Kincaid and Gunning Fog Scores standards trying to keep them low for better legibility.

Just reformulate it. Don't add initial phrase.

When necessary, you can try to syntetize it always keeping the initial meaning.

<<SYS>>
Original question: {question}
Original Answer: {original_answer} [/INST]
'''

        # Retry automatico
        retries = 3
        while retries > 0:
            try:
                # Chiamata API senza stream, più veloce
                completion = client.chat.completions.create(
                    model="writer/palmyra-med-70b-32k",
                    #model="mistralai/mixtral-8x22b-instruct-v0.1",
                    messages=[{"role": "user", "content": prompt_template}],
                    temperature=0.2,
                    top_p=0.7,
                    max_tokens=1024,
                    stream=False  # disattiva stream
                )

                rewritten_answer = completion.choices[0].message.content.strip()
                break  # esce dal retry loop se va bene

            except Exception as e:
                print(f"Retrying... Remaining attempts: {retries-1}")
                retries -= 1
                time.sleep(5)  # aspetta un po' prima di riprovare
                if retries == 0:
                    raise e  # rilancia l'eccezione se finiti i tentativi

        # Aggiorna il DataFrame
        update_dataframe(index, rewritten_answer)

        # Stampa output semplice per monitorare
        print(f"Index {index}: {rewritten_answer}\n")

        count += 1

        # Salvataggio ogni `save_every`
        # Salvataggio ogni `save_every`
        if count % save_every == 0:
            df.to_csv(output_path, index=False)
            print(f"🔵 Autosave at index {index} - Data saved to {output_path}")
            time.sleep(1)
            clean_console()

    except Exception as e:
        print(f"❌ Error processing index {index}: {e}")

# Salvataggio finale
df.to_csv(output_path, index=False)
print(f"✅ Processing completed. Final data saved to {output_path}")
```

```python
import pandas as pd

# Funzione per creare df3
def crea_df3_selettivo(df, df1):
    # Crea una copia di df1
    df3 = df1.copy()

    # Identifica gli indici delle righe da aggiornare
    indici_da_sostituire = df1.index

    # Sostituisci i valori di 'question' con quelli di df
    df3.loc[indici_da_sostituire, 'question'] = df.loc[indici_da_sostituire, 'question']

    # Azzera i valori delle altre colonne solo per le righe modificate
    for col in df3.columns:
        if col != 'question':
            df3.loc[indici_da_sostituire, col] = '-'

    return df3

# Esempio di utilizzo
df3 = crea_df3_selettivo(df1, df1)
print(df3)

```

```python
from openai import OpenAI
import pandas as pd

def update_dataframe(query, Risposta, row_index):
    """
    Aggiorna il DataFrame `df3` alla riga specificata con la risposta fornita.
    """
    global df3  # Assicurati che il DataFrame `df3` esista nello scope globale
    answer = Risposta.strip()  # La risposta diventa il contenuto aggiornato della colonna `answer`
    df3.at[row_index, 'answer'] = answer  # Aggiorna la riga specificata con la risposta

# Configurazione del client OpenAI
client = OpenAI(
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=""
)

# Itera solo sulle righe di `df3` che hanno un '-' nella colonna `answer`
rows_with_dash = df3[df3['answer'] == '-'].index  # Ottieni gli indici delle righe con '-'

for index in rows_with_dash:
    query = df3.loc[index, 'question']  # Prendi la query dalla colonna `question`

    # Costruisci il template del prompt
    prompt_template = f'''[INST] <<SYS>>

    You are a helpful, respectful, and honest Doctor. Always answer as helpfully as possible using the context text provided.

    Answer describing in a scientific way. Be formal during the answer. Use the third person.

    Answer like a doctor who is answering on an online forum or who is speaking with a patient.

    Don't make a list. Answer in a discursive way.

    Don't enumerate. Answer being polite and caring.

    Don't use structures like "in summary" or "following" or "In addition."

    <<SYS>>

    Question: {query} [/INST]
    '''
    try:
        # Chiamata API per generare la risposta
        completion = client.chat.completions.create(
            model="mistralai/mixtral-8x22b-instruct-v0.1",
            messages=[{"role": "user", "content": prompt_template}],
            temperature=0.5,
            top_p=1,
            max_tokens=1024,
            stream=True
        )

        # Raccogli la risposta generata
        Risposta = ""
        for chunk in completion:
            if chunk.choices[0].delta.content:
                Risposta += chunk.choices[0].delta.content

        # Debugging: Stampa la risposta per verificare
        print(f"Response for index {index}: {Risposta}")

        # Aggiorna il DataFrame `df3` con la risposta generata
        update_dataframe(query, Risposta, index)

    except Exception as e:
        print(f"Error during API call for index {index}: {e}")
```
